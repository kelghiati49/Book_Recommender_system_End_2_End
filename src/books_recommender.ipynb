### preprocessing of ratings dataset

```python
# let's explore our dataset
ratings['user_id'].value_counts()
```

here we can see that some user rate more than 200 books, other rated less (1 or 2 , just like me hhh)

```python
x = ratings["user_id"].value_counts() > 200
print("users how rate more than 200 book:", x[x].shape)
```

so let's keep users how rated more than 200 books

```python
# let's keep users with more than 200 book rating
y = x[x].index
ratings = ratings[ratings['user_id'].isin(y)]
ratings.head()
```

```python
ratings.shape
```

```python
# Now we'll filter books that have more than 200 ratings
ratings_with_books = ratings.merge(books, on ="ISBN")
ratings_with_books.head()
```

```python
ratings_with_books.shape
```

```python
# for each book how much is rated in the dataset
num_ratings = ratings_with_books.groupby('title')['rating'].count().reset_index()
num_ratings.head()
```

* data inbalance : some books are rated one time other more than 200 times.
=> so we should take only books that have at least more than 50 ratings.


```python
num_ratings.rename(columns = {
    "rating":"num_of_ratings"
}, inplace = True)
num_ratings.head(2)
```

```python
# now let's merge aagain this information with the previous ones
final_ratings=ratings_with_books.merge(num_ratings, on="title")
final_ratings.head(2)
```

```python
final_ratings.shape
```

```python
# let's filter books with more than 50 ratings
final_ratings = final_ratings[final_ratings['num_of_ratings'] >= 50]
final_ratings.head()
```

```python
final_ratings.shape
```

```python
# let's remove duplicate values
final_ratings.drop_duplicates(['user_id','title'], inplace =True)
final_ratings.shape
```

```python
# and here's our final dataset lol
final_ratings
```

### Convert To Pivot Table

Why converting into a pivot table  
- clustering algorithms expect input shaped into : matrix, no text, no duplicates, no multiple rows per entity.  
- books with similar rating patterns -> cluster together  
- Users with similar taste -> cluster together

```python
book_pivot = final_ratings.pivot_table(columns='user_id', index='title', values='rating')
book_pivot.fillna(0, inplace=True)
book_pivot
```

```python
from scipy.sparse import csr_matrix
book_sparse = csr_matrix(book_pivot) # automaticly ignore 0 in the dataset 
                                    # and consider other values
                                    #=> model became much faster
```

### Model Building

```python
from sklearn.neighbors import NearestNeighbors

model = NearestNeighbors(algorithm = "brute")
model.fit(book_sparse)
```

### test the model

```python
# try to get the suggestion of the user i (ex: 254)
distance, suggestion = model.kneighbors(book_pivot.iloc[254,:].values.reshape(1,-1),n_neighbors=6)
```

```python
distance
```

```python
suggestion
```

```python
# some books the user 254 might like also :
for i in range(len(suggestion)):
    print(book_pivot.index[suggestion[i]])
```
